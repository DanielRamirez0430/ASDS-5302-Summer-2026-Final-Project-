## Problem 7

```python
bins = [0, 3, 6, np.inf]
labels = ['poor', 'medium', 'good']

sample['quality_level'] = pd.cut(sample['quality'], bins = bins, labels = labels)

print(sample[['quality', 'quality_level']].head(10))
```

```python
# For this problem i created a new categorical column called quality_level by using
# pd.cut to split the numerical quality ratings into bins. Then i printed the first
# 10 rows of the df to make sure quality had the correct bin in quality_level. 
```

## Problem 8

```python
plt.figure(figsize = (6,5))
sns.set_theme(style = 'whitegrid')

sns.histplot(data = sample, x = 'quality_level', discrete = True, color = 'purple')

plt.title('Histogram of Wine Quality Levels')
plt.xlabel('Quality Level')
plt.ylabel('Count')

plt.show()
```

```python
# I built a histogram of the new quality_level using sns.histplot from the 
# seaborn library, and I set discrete=True so each category gets its own bar.
```

## Problem 9

```python
sample['alcohol_level'] = pd.cut(sample['alcohol'], bins = [-np.inf, 10, 12, np.inf], labels = ['low_alcohol', 'medium_alcohol', 'high_alcohol'])

sample['acidity_level'] = np.where(sample['fixed acidity'] > 8.0, 'high_acidity', 'low_acidity')

grouped_quality = sample.groupby(['alcohol_level', 'acidity_level'], observed = False)['quality'].mean()

print(grouped_quality)
```

```python
# For this problem I created a two new categorical features by binning alcohol into
# three parts using pd.cut and then splitting them into high or low using np.where.
# Then i grouped the sample by both new columns to calculate and print the average
# quality score for each combination.
```

## Problem 10

```python
plt.figure(figsize = (7,5))
sns.set_theme(style = "whitegrid")

sns.scatterplot(data = sample, x = 'alcohol', y = 'quality', alpha = 0.6, color = 'purple')

plt.title('Relationship Between Alcohol Content and Wine Quality')
plt.xlabel('Alcohol Content (%)')
plt.ylabel('Wine Quality')

plt.yticks(range(int(sample['quality'].min()), int(sample['quality'].max()) + 1))

plt.show()
```

```python
# For this problem i created a scatterplot using seaborn to look at the relationship
# between alcohol content and wine quality. We can see a slight upward trend showing that
# wines with higher alcohol content are more likely to achieve quality ratings of 7 or 8.
```

## Problem 11

```python
numerical_cols = df_wine.select_dtypes(include = ['float64', 'int64']).columns

plt.figure(figsize = (10, 12))

for i, col in enumerate(numerical_cols, 1): 
    plt.subplot(4, 3, i)
    sns.boxplot(x = df_wine[col], color = 'purple')
    plt.title(col)

plt.tight_layout()
plt.show()
```

```python
# Yes outliers can be seen from the grid of boxplots i made for all the numeric
# columns in df_wine. We can catch the outliers using the IQR rule or z-scores, 
# and they can be handled in various ways, like dropping them or using a log
# transform.
```

## Problem 12

```python
corr_matrix = df_wine.select_dtypes(include = ['float64', 'int64']).corr()

plt.figure(figsize = (9,7))
sns.heatmap(corr_matrix, annot = True, fmt = ".2f", cmap = 'coolwarm', vmin = -1, vmax = 1)
plt.title("Correlation Matrix of Wine Feature")
plt.show()
```

```python
# For this problem i calculated the corr matrix for all of the numeric features
# in df_wine and plotted it using seaborn heatmap. This makes it easy to spot highly
# correlated features like free sulfur dioxide and total sulfur dioxide. We can 
# deal with it by either dropping redundant features or using PCA to merge collinear features.
```
